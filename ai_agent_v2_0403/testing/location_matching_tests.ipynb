```python
# 🔧 Normalization utility
def normalize_location(loc: str) -> str:
    if not loc:
        return ""
    loc = loc.lower().strip()
    for word in ["airport", "city", "usa", ",", ".", "  "]:
        loc = loc.replace(word, "")
    return loc.title().strip()
```

```python
# 🧠 Fuzzy match
from difflib import get_close_matches

def fuzzy_lookup(key: str, candidates: list[str], cutoff=0.6) -> str:
    matches = get_close_matches(key, candidates, n=1, cutoff=cutoff)
    return matches[0] if matches else None
```

```python
# 🔍 Try real edge case lookups
raw_input = "New York"
normalized = normalize_location(raw_input)
fuzzy_match = fuzzy_lookup(normalized, distances.keys())

print("Original:", raw_input)
print("Normalized:", normalized)
print("Fuzzy Match:", fuzzy_match)
print("Matched Entry:", distances.get(fuzzy_match))
```

```python
# 🔁 Compare: no normalization vs full normalization
def test_lookup(input_loc):
    print(f"🔍 Testing lookup for: {input_loc}")
    direct = distances.get(input_loc)
    norm = normalize_location(input_loc)
    fuzzy = fuzzy_lookup(norm, distances.keys())
    result = distances.get(fuzzy) if fuzzy else None

    print("Direct Match:", "✅" if direct else "❌")
    print("Normalized Input:", norm)
    print("Fuzzy Matched:", fuzzy if fuzzy else "❌ No Match")
    return result

# Test different inputs
test_lookup("New York")
test_lookup("new york city")
test_lookup("DFW airport")
test_lookup("dallas")
test_lookup("JFK airport")
```
