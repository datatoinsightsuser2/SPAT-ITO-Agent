# This function works for any from and to location.

```python
import os
import requests
def get_realtime_distance(from_location, to_location):
    MILES_IN_METER = 0.000621371
    if not from_location or not to_location:
        return 100  # default fallback distance

    headers = {
        'Content-Type': 'application/json',
        'X-Goog-Api-Key': 'AIzaSyATwfZbWvLDnlWWj4-5shMPPcV7_BOUfx4',#os.getenv("GOOGLE_API_KEY"), this is only for testing. We'll keep it in our env
        'X-Goog-FieldMask': 'routes.distanceMeters',
    }

    json_data = {
        'origin': {'address': from_location},
        'destination': {'address': to_location},
        'travelMode': 'DRIVE',
        'languageCode': 'en-US',
    }

    try:
        response = requests.post(
            'https://routes.googleapis.com/directions/v2:computeRoutes',
            headers=headers,
            json=json_data
        )
        if response.status_code == 200:
            distance_meters = response.json()["routes"][0]["distanceMeters"]
            return round(distance_meters * MILES_IN_METER)
        else:
            print(f"API Error: {response.status_code}")
            return 100
    except Exception as e:
        print(f"API Call Failed: {e}")
        return 100
```
