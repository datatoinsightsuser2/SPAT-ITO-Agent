```python
def test_map_rate_with_route(from_airport, to_airport, charged_kg, rates_data):
    for rate_entry in rates_data:
        if rate_entry["fromAirport"] == from_airport and rate_entry["toAirport"] == to_airport:
            tiered = rate_entry["rates"]
            cost = map_rate(charged_kg, tiered)
            return {
                "from": from_airport,
                "to": to_airport,
                "charged_kg": charged_kg,
                "rate_applied": cost,
                "airlineId": rate_entry["airlineId"]
            }
    return {"error": "No route found in rates.json"}

# Example usage
charged_kg = 100
from_airport = "ORD"
to_airport = "HYD"

charged_kg = 100
from_airport = "JFK"
to_airport = "BLR"
# Load your rates.json file
import json
with open("data/rates_updated_full.json", "r") as f:
    rates_data = json.load(f)

# Call the test
result = test_map_rate_with_route(from_airport, to_airport, charged_kg, rates_data)
print(result)
```
